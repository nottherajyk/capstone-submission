# 04. Operational Label Validation & Class Diagnostics

## Methodological Framing
The label represents a genuine operational review opportunity. Rather than vague 'decaying trends', we define deterioration strictly across a future horizon $[T+1, T+H]$ and enforce explicit exclusion criteria for unrepresentative or sparse history records.

### Objectives:
1. Define binary deterioration target: $\ge 20\%$ click drop or $\ge 2.0$ position loss in future window.
2. Filter ineligible records based on minimum historical impression and click thresholds.
3. Report live label diagnostics: eligible count, positive count, negative count, and base rate.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

root_dir = Path.cwd().parent if Path.cwd().name == "work" else Path.cwd()
sys.path.insert(0, str(root_dir))

from src.config import load_config
from src.data import load_dataset
from src.labels import construct_operational_label

config = load_config()
try:
    df = load_dataset(config.raw_data_path, config)
    is_eligible, y_target, diag = construct_operational_label(df, config.label)
    
    print("=========================================")
    print("OPERATIONAL LABEL DIAGNOSTICS (LIVE RUN)")
    print("=========================================")
    print(f"Total Records:       {diag.total_population:,}")
    print(f"Eligible Records:    {diag.eligible_population:,}")
    print(f"Excluded Records:    {diag.excluded_count:,}")
    print(f"Positive (Decay):    {diag.positive_count:,}")
    print(f"Negative (Stable):   {diag.negative_count:,}")
    print(f"Observed Base Rate:  {diag.positive_rate * 100:.2f}%")
    print("Exclusion Reasons:", diag.exclusion_breakdown)
except FileNotFoundError:
    print("[STATUS: AWAITING REAL DATA EXECUTION] - Label diagnostics will compute on connected warehouse data.")
```
